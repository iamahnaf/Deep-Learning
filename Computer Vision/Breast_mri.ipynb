```python
import tensorflow as tf
import os

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# Create optimized datasets (images remain unscaled in 0-255 range here; model will handle scaling)
train_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(BASE_PATH, 'train'),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary',
    color_mode='rgb',
    shuffle=True
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(BASE_PATH, 'val'),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary',
    color_mode='rgb',
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(BASE_PATH, 'test'),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary',
    color_mode='rgb',
    shuffle=False
)

# Configure performance optimization with prefetching
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)

print("Optimized tf.data Datasets ready!")
```

```python
# Define the data augmentation sequential block
data_augmentation = Sequential([
    tf.keras.layers.Rescaling(1./255), # Rescale pixel values [0, 255] to [0.0, 1.0]
    tf.keras.layers.RandomRotation(0.027),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomTranslation(height_factor=0.05, width_factor=0.05),
    tf.keras.layers.RandomFlip("horizontal")
])

# Building the CNN with Integrated Augmentation Layers
CNN = Sequential([
    # Input layer specification
    tf.keras.layers.Input(shape=(224, 224, 3)),

    # Augmentation & Rescaling layers (Only active during training, auto-bypassed during validation/testing)
    data_augmentation,

    # Feature Extraction Convolutional Blocks
    Conv2D(32, (3, 3), activation='relu'),
    MaxPool2D(2, 2),

    Conv2D(64, (3, 3), activation='relu'),
    MaxPool2D(2, 2),

    Conv2D(128, (3, 3), activation='relu'),
    MaxPool2D(2, 2),

    # Fully Connected Dense Classifier
    Flatten(),
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

CNN.summary()
```

```python
CNN.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy',tf.keras.metrics.AUC(name='auc')] #auc for medical
)
```

```python
callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('best_breast_mri_model.keras', monitor='val_auc', save_best_only=True, mode='max'),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, verbose=1)
]

# Start high-speed training on pre-fetched TF datasets with native GPU augmentations!
history = CNN.fit(
    train_ds,
    epochs=5,
    validation_data=val_ds,
    callbacks=callbacks
)
```

```python
# Plot training curves and evaluate tumor classification on the test set
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(history.history['loss'], label='Training loss')
axes[0].plot(history.history['val_loss'], label='Validation loss')
axes[0].set_title('Loss curve')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(history.history['accuracy'], label='Training accuracy')
axes[1].plot(history.history['val_accuracy'], label='Validation accuracy')
axes[1].set_title('Accuracy curve')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(alpha=0.3)

axes[2].plot(history.history['auc'], label='Training AUC')
axes[2].plot(history.history['val_auc'], label='Validation AUC')
axes[2].set_title('AUC curve')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('AUC')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Generate predictions for every test image
true_labels = np.concatenate([labels.numpy().ravel() for _, labels in test_ds]).astype(int)
predicted_probabilities = CNN.predict(test_ds, verbose=1).ravel()
predicted_labels = (predicted_probabilities >= 0.5).astype(int)

confusion = confusion_matrix(true_labels, predicted_labels)
plt.figure(figsize=(6, 5))
sns.heatmap(
    confusion,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=classes,
    yticklabels=classes
)
plt.title('Tumor classification confusion matrix')
plt.xlabel('Predicted class')
plt.ylabel('True class')
plt.tight_layout()
plt.show()

print(classification_report(true_labels, predicted_labels, target_names=classes))
print(f"Test accuracy: {accuracy_score(true_labels, predicted_labels):.4f}")
```

```python
# Grad-CAM: inspect which image regions influenced the model's prediction
last_conv_layer = next(
    layer for layer in reversed(CNN.layers)
    if isinstance(layer, tf.keras.layers.Conv2D)
)

# Use the first test image. Change image_index to inspect another image.
image_index = 0
image_batch, label_batch = next(iter(test_ds))
image = image_batch[image_index:image_index + 1]
true_label = int(label_batch.numpy().ravel()[image_index])

with tf.GradientTape() as tape:
    layer_output = image
    conv_outputs = None
    for layer in CNN.layers:
        layer_output = layer(layer_output, training=False)
        if layer is last_conv_layer:
            conv_outputs = layer_output
    prediction = layer_output
    malignant_probability = prediction[:, 0]
    predicted_label = tf.cast(malignant_probability >= 0.5, tf.int32)
    target_score = tf.where(
        predicted_label == 1,
        malignant_probability,
        1.0 - malignant_probability
    )

grads = tape.gradient(target_score, conv_outputs)
pooled_grads = tf.reduce_mean(grads, axis=(1, 2))
conv_outputs = conv_outputs[0]
pooled_grads = pooled_grads[0]
heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
heatmap = tf.maximum(heatmap, 0)
heatmap /= tf.maximum(tf.reduce_max(heatmap), tf.keras.backend.epsilon())
heatmap = heatmap.numpy()

# Resize the heatmap to the image size and keep only the strongest activations.
original_image = image[0].numpy().astype('uint8')
heatmap_resized = tf.image.resize(
    heatmap[..., np.newaxis], IMG_SIZE, method='bilinear'
).numpy().squeeze()
heatmap_resized = np.clip(heatmap_resized, 0, 1)
activation_threshold = np.percentile(heatmap_resized, 85)
important_regions = heatmap_resized >= activation_threshold

predicted_label = int(predicted_label[0].numpy())
malignant_probability = float(malignant_probability[0].numpy())
print(f"True class: {classes[true_label]}")
print(f"Predicted class: {classes[predicted_label]}")
print(f"Malignant probability: {malignant_probability:.4f}")
print(f"Explained class: {classes[predicted_label]}")
print(f"Important-region threshold: {activation_threshold:.2f}")
print(f"Grad-CAM layer: {last_conv_layer.name}")

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes[0, 0].imshow(original_image, cmap='gray')
axes[0, 0].set_title('Original breast MRI')

heatmap_plot = axes[0, 1].imshow(heatmap_resized, cmap='jet', vmin=0, vmax=1)
axes[0, 1].set_title('Feature importance heatmap')
fig.colorbar(heatmap_plot, ax=axes[0, 1], fraction=0.046, pad=0.04)

axes[0, 2].imshow(original_image, cmap='gray')
axes[0, 2].imshow(heatmap_resized, cmap='jet', alpha=0.45, vmin=0, vmax=1)
axes[0, 2].set_title('Transparent importance overlay')

axes[1, 0].imshow(original_image, cmap='gray')
axes[1, 0].imshow(important_regions, cmap='Reds', alpha=0.65, vmin=0, vmax=1)
axes[1, 0].set_title('Strongest 15% of activations')

axes[1, 1].imshow(original_image, cmap='gray')
axes[1, 1].contour(important_regions, levels=[0.5], colors='yellow', linewidths=1.5)
axes[1, 1].set_title('Important-region contour')

axes[1, 2].axis('off')
axes[1, 2].text(
    0.05, 0.85,
    'Interpretation:\n\n'
    'Red/yellow areas show image regions\n'
    'that influenced the prediction most.\n\n'
    'They are model features, not a\n'
    'verified tumor boundary.',
    fontsize=12,
    va='top'
)

for axis in axes.flat:
    axis.axis('off')
axes[1, 2].axis('off')
plt.suptitle(
    f"Grad-CAM explanation for predicted class: {classes[predicted_label]}",
    fontsize=14
)
plt.tight_layout()
plt.show()
```

```python
# VGG16 transfer-learning model for comparison with the custom CNN
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.layers import GlobalAveragePooling2D, Lambda

vgg_augmentation = Sequential([
    tf.keras.layers.RandomRotation(0.027),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomTranslation(height_factor=0.05, width_factor=0.05),
    tf.keras.layers.RandomFlip('horizontal')
], name='vgg_augmentation')

vgg_base = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(*IMG_SIZE, 3)
)
vgg_base.trainable = False

VGG_CNN = Sequential([
    tf.keras.layers.Input(shape=(*IMG_SIZE, 3)),
    vgg_augmentation,
    Lambda(preprocess_input, name='vgg16_preprocess'),
    vgg_base,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
], name='vgg16_breast_mri_classifier')

VGG_CNN.summary()
```

```python
VGG_CNN.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
)

vgg_callbacks = [
    EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True),
    ModelCheckpoint(
        'best_vgg16_breast_mri_model.keras',
        monitor='val_auc',
        save_best_only=True,
        mode='max'
    ),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)
]

vgg_history = VGG_CNN.fit(
    train_ds,
    epochs=10,
    validation_data=val_ds,
    callbacks=vgg_callbacks
)
```

```python
# Compare custom CNN and VGG16 on the same test dataset
vgg_test_loss, vgg_test_accuracy, vgg_test_auc = VGG_CNN.evaluate(test_ds, verbose=1)

print('Custom CNN test metrics are shown above in the earlier evaluation cell.')
print(f'VGG16 test accuracy: {vgg_test_accuracy:.4f}')
print(f'VGG16 test AUC: {vgg_test_auc:.4f}')

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(history.history['val_loss'], label='Custom CNN')
axes[0].plot(vgg_history.history['val_loss'], label='VGG16')
axes[0].set_title('Validation loss comparison')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(history.history['val_accuracy'], label='Custom CNN')
axes[1].plot(vgg_history.history['val_accuracy'], label='VGG16')
axes[1].set_title('Validation accuracy comparison')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(alpha=0.3)

axes[2].plot(history.history['val_auc'], label='Custom CNN')
axes[2].plot(vgg_history.history['val_auc'], label='VGG16')
axes[2].set_title('Validation AUC comparison')
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('AUC')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ResNet50 transfer-learning model for comparison
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet_preprocess_input

resnet_augmentation = Sequential([
    tf.keras.layers.RandomRotation(0.027),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomTranslation(height_factor=0.05, width_factor=0.05),
    tf.keras.layers.RandomFlip('horizontal')
], name='resnet_augmentation')

resnet_base = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(*IMG_SIZE, 3)
)
resnet_base.trainable = False

RESNET_CNN = Sequential([
    tf.keras.layers.Input(shape=(*IMG_SIZE, 3)),
    resnet_augmentation,
    Lambda(resnet_preprocess_input, name='resnet50_preprocess'),
    resnet_base,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
], name='resnet50_breast_mri_classifier')

RESNET_CNN.summary()
```

```python
RESNET_CNN.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.AUC(name='auc')]
)

resnet_callbacks = [
    EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True),
    ModelCheckpoint(
        'best_resnet50_breast_mri_model.keras',
        monitor='val_auc',
        save_best_only=True,
        mode='max'
    ),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)
]

resnet_history = RESNET_CNN.fit(
    train_ds,
    epochs=10,
    validation_data=val_ds,
    callbacks=resnet_callbacks
)
```
